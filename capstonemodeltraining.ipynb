```python
import pandas as pd


mouse=pd.read_csv("datasets/mouse_movement_dataset.csv")
print(mouse.tail())
```

```python
import joblib

from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)


mousex=mouse.drop(columns=["session_id","label"])
mousey=mouse["label"]

mousextrain,mousextest,mouseytrain,mouseytest=train_test_split(mousex,mousey,test_size=0.1,random_state=42,stratify=mousey)
```

```python
from catboost import CatBoostClassifier

model = CatBoostClassifier(
    iterations=500,
    learning_rate=0.05,
    depth=6,
    loss_function="Logloss",
    auto_class_weights="Balanced",
    random_seed=42,
    verbose=0,
    thread_count=1,
    allow_writing_files=False
)
```

```python
model.fit(mousextrain,mouseytrain)

pred=model.predict(mousextest)

accuracy_score(mouseytest,pred)
```

```python
joblib.dump(model,"mousemodel.pkl")
```

```python
from catboost import CatBoostClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

sessionx=session.drop(columns=["ROBOT"])
sessiony=session["ROBOT"]

sessionxtrain,sessionxtest,sessionytrain,sessionytest=train_test_split(sessionx,sessiony,test_size=0.1,random_state=42,stratify=sessiony)

sessionmodel = CatBoostClassifier(
    iterations=500,
    learning_rate=0.05,
    depth=6,
    loss_function="Logloss",
    auto_class_weights="Balanced",
    random_seed=42,
    verbose=0,
    thread_count=1,
    allow_writing_files=False
)

sessionmodel.fit(sessionxtrain,sessionytrain)
pred=sessionmodel.predict(sessionxtest)
print(accuracy_score(sessionytest,pred))
```

```python
joblib.dump(sessionmodel,"sessionmodel.pkl")
```
