```python
import pandas as pd


mouse=pd.read_csv("mouse_movement_dataset.csv")
print(mouse.tail())
```

```python
import joblib

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)


mousex=mouse.drop(columns=["session_id","label"])
mousey=mouse["label"]

mousextrain,mousextest,mouseytrain,mouseytest=train_test_split(mousex,mousey,test_size=0.1,random_state=42,stratify=mousey)
```

```python
model = RandomForestClassifier(n_estimators=100,max_depth=4,min_samples_split=5,min_samples_leaf=3,max_features="sqrt",class_weight="balanced",random_state=42,n_jobs=-1)
```

```python
model.fit(mousextrain,mouseytrain)

pred=model.predict(mousextest)

accuracy_score(mouseytest,pred)
```

```python
joblib.dump(model,"mousemodel.pkl")
```

```python
session=pd.read_csv("session.csv")
print(session.head())
```

```python
sessionx=session.drop(columns=["ROBOT"])
sessiony=session["ROBOT"]

sessionxtrain,sessionxtest,sessionytrain,sessionytest=train_test_split(sessionx,sessiony,test_size=0.1,random_state=42,stratify=sessiony)

sessionmodel=RandomForestClassifier(
    n_estimators=300,
    max_depth=15,
    min_samples_split=10,
    min_samples_leaf=4,
    max_features="sqrt",
    class_weight="balanced_subsample",
    bootstrap=True,
    random_state=42,
    n_jobs=-1
)

sessionmodel.fit(sessionxtrain,sessionytrain)
pred=sessionmodel.predict(sessionxtest)
print(accuracy_score(sessionytest,pred))
```

```python
joblib.dump(sessionmodel,"sessionmodel.pkl")
```
